This notebook is for feature engineering required for processing the data used for modeling.

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('../data/processed.csv')
df.head()
```

```python
tenure_bins = [0, 6, 12, 24, np.inf]
labels = ['0-6', '6-12', '12-24', '24+']

df['Tenure_Bucket'] = pd.cut(df['Tenure in Months'], bins=tenure_bins, labels=labels, right=False)
df[['Tenure in Months', 'Tenure_Bucket']].head(10)
```

```python
df['Value_Segment'] = pd.qcut(df['CLTV'], q=3, labels=['Low', 'Medium', 'High'])

df[['CLTV', 'Value_Segment']].head(10)
```

```python
df['Revenue_per_Month'] = df['Total Revenue'] / df['Tenure in Months']

df['Revenue_per_Month'] = df['Revenue_per_Month'].replace([float('inf')], 0)
df['Revenue_per_Month'] = df['Revenue_per_Month'].fillna(0)
```

```python
df['Revenue_Deviation'] = df['Revenue_per_Month'] / df['Monthly Charge']

df[['Total Revenue', 'CLTV', 'Monthly Charge', 'Revenue_per_Month', 'Revenue_Deviation']].head(10)
```

```python
cols_to_drop = ['Under 30', 'Senior Citizen', 'Quarter', 'Country', 'State', 'City', 'Zip Code', 'Churn Reason', 'Churn Score', 'CLTV', 'Customer Status']
df = df.drop(columns=[col for col in cols_to_drop if col in df.columns])
df = df.drop(['Customer ID'], axis=1)
```

```python
df.to_csv('../data/processed.csv')
```
